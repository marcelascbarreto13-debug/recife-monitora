# Recife Monitora — Preparação dos Dados

## Objetivo

Esta etapa tem como objetivo tratar os problemas de qualidade identificados durante o entendimento dos dados, preservando as regras de negócio e mantendo os dados originais inalterados.

Os tratamentos serão realizados de forma documentada e reproduzível, com validações antes e depois das alterações.

**Importante:** a base utilizada neste projeto é sintética e não representa resultados reais das equipes ou unidades da Secretaria de Saúde do Recife.


```python
import pandas as pd
from pathlib import Path

# Caminho para a pasta do projeto
PASTA_PROJETO = Path.cwd().parent

# Carregar os arquivos originais
equipes = pd.read_csv(PASTA_PROJETO / "dados" / "equipes.csv")
resultados = pd.read_csv(PASTA_PROJETO / "dados" / "resultados_ciclos.csv")
eixo3 = pd.read_csv(PASTA_PROJETO / "dados" / "eixo3_indicadores.csv")
nps = pd.read_csv(PASTA_PROJETO / "dados" / "nps_atributos.csv")
planos = pd.read_csv(PASTA_PROJETO / "dados" / "planos_acao.csv")

print("Tabelas carregadas com sucesso!")
print("Equipes:", equipes.shape)
print("Resultados:", resultados.shape)
print("Eixo 3:", eixo3.shape)
print("NPS:", nps.shape)
print("Planos de ação:", planos.shape)

# Criar cópias para trabalhar com segurança
equipes_trat = equipes.copy()
resultados_trat = resultados.copy()
eixo3_trat = eixo3.copy()
nps_trat = nps.copy()
planos_trat = planos.copy()

print("Cópias de trabalho criadas!")

print("Equipes:", equipes_trat.shape)
print("Resultados:", resultados_trat.shape)
print("Eixo 3:", eixo3_trat.shape)


# Conferir os tipos de equipe antes do tratamento
print(equipes_trat["tipo_equipe"].value_counts(dropna=False).sort_index())
```
